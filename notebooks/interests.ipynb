```python
def compound(principal: float, rate: float, increment: float, nmonths: int) -> float:
    """
    https://moneysmart.gov.au/saving/compound-interest

    principal - the amount in the account at the beginning
    rate - per annum interest rate in percentage e.g. 2.5
    increment - monthly incomes into the account from other means (independent of the monthly interest paid)
    nmonths - number of months

    the equation is: x = p x (1 + r) ^ n
    p is the principal
    r is the rate of interest - for the interval interest is being paid (p.a. rate / 12)
    n - number of interest paid intervals (i.e. months)

    however, this equation does not take into account external incomes??
    """

    balance = principal
    total_interest = 0.000 # total compound interest earned
    
    for _ in range(nmonths): # for each compound interest paid interval
        # interest is paid at the end of the month - so the amount will be the principal at the beginning of the month + everything that's come into that account during that month
        interest = (balance + increment) * (rate / 100 / 12.000) # rate / 100 gives the fraction - 2.1 becomes 0.021
        # print(interest)
        total_interest += interest
        balance += (increment + interest) # balance at the end of the month
        
    return (balance, total_interest)
```
