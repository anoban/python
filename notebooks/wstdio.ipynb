```python
cppfiles = [source for source in os.listdir(r"./../../cpp20/") if (source.endswith(".cpp") or source.endswith(".hpp"))]
```

```python
contents = dict.fromkeys(cppfiles)

for file in cppfiles:
    with open(file=f"./../../cpp20/{file}", mode="rt") as fp:
        contents[file] = fp.read()
```
